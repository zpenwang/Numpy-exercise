## Task 4: Simulating a dataset

The famous Mincer regression has the form

$ln~w = \beta_0 + \beta_1 s + \beta_2 x + \beta_3 x^2 + \epsilon$

Where:

- w is the wage
- s is years of schooling
- x is years of experience
- $\epsilon$ is an unobservable term

And the betas are coefficients of a linear regression model. Assume parameter values:

- $\beta_0$ = 7
- $\beta_1$ = 0.1
- $\beta_2$ = 0.05
- $\beta_3$ = 0.01

Your goal is to create a simulated dataset for estimating such an equation. I.e. you
want an X matrix with the four columns intercept, schooling, experience and
experience squared and a Y vector containing the log wage. Call the matrix of regressors `exog`  and the Y vector `endog`.

The following assumptions are not realistic; We want to make your lives easy. 

Assume that years of schooling is a discrete variable that takes the values 9, 10, 12
15 and 17 with equal probabilities. Years of experience is a continuous variable that 
follows a uniform distribution between 0 and 40. The unobservable term follows a 
normal distribution with mean 0 and standard deviation 1.5. All random variables are 
independent. 

You might want to google how to concatenate numpy arrays. Use a matrix-vector
multiplication to calculate the dependent variable.

Simulate 500 individuals



```python
np.random.default_rng(42)
s = np.random.choice([9,10,12,15,17], size = 500)
x = np.random.uniform(0, 40, size = 500)
epsilon = np.random.normal(0, 1.5, size = 500)

intercept = np.ones(500)
experience_squared = x ** 2
exog = np.column_stack((intercept,s, x, experience_squared))

beta = np.array([7, 0.1, 0.5, -0.01])
endog = exog @ beta + epsilon
```

## Task 6: Implement a naive OLS estimator

The OLS estimator is given by:

$\hat{\beta} = (X'X)^{-1}X'y$

Implement a function called `reg(exog, endog)` that implements this formula. 

Evaluate it on the simulated dataset. If you want, you can increase the number of 
simulated individuals until you get really close to the true beta vector.

**Note**: This is for you to practice using numpy. Never use this for an actual
regression! There are much more robust ways of calculating an OLS estimator and they 
are implemented for you in statistics packages. They don't use matrix inverses, 
for [reasons](https://www.johndcook.com/blog/2010/01/19/dont-invert-that-matrix/).

```python
def reg(exog, endog):
    beta_hat = np.linalg.inv(exog.T @ exog) @ exog.T @ endog
    return beta_hat
```

```python
beta_hat = reg(exog, endog)

print(beta_hat)
```

## Task 7: Broadcasting (Bonus Question)

Assume we have a Cobb-Douglas production function given by:

$y = x_1^{\alpha} \cdot x_2^{1 - \alpha}$

For a structural model we want to evaluate this function on a Cartesian grid of possible 
values for $x_1$ and $x_2$. The individual grids are defined below. 

Implement the function `cartesian_cobb_douglas` that takes `x1_grid`, `x2_grid` and
`alpha` as arguments and returns a 2d array of function evaluations. If `x1_grid` has
length `n` and `x2_grid` has length `m`, the matrix of function evaluations should be
`n` × `m`.

Use broadcasting to do this. Do not use a loop. 

```python
x1_grid = np.linspace(0, 10, 5)
x2_grid = np.linspace(1, 20, 3)
alpha = 0.5
print(x1_grid)
print(x2_grid)
```

```python
def cartesian_cobb_douglas(x1_grid, x2_grid, alpha):
    y = x1_grid.reshape(-1,1) ** alpha * x2_grid.reshape(1,-1) ** (1 - alpha)
    return y
```

```python
result = cartesian_cobb_douglas(x1_grid, x2_grid, alpha)

print(result)
print(result.shape)
```

The inputs are tiny on purpose, so you can easily verify that everything works. 

`output[2, 2]` should be 10.0 as you can easily verify. 

If you want, you can make the grid much finer. Doing the calculation will still be very
fast because the entire calculation is actually done in a compiled language.
